```python
import pyttsx3

poem = """A whisper rides the morning breeze,
Through dancing leaves on quiet trees.
The sun peeks out with golden grace,
And paints a smile on nature's face.
In silence, peace begins to grow."""

# Initialize TTS engine
engine = pyttsx3.init()

# Adjust rate and volume
engine.setProperty('rate', 115)
engine.setProperty('volume', 1.0)

# Set voice (choose English Female if available)
voices = engine.getProperty('voices')
for voice in voices:
    if "female" in voice.name.lower() and "english" in voice.name.lower():
        engine.setProperty('voice', voice.id)
        break

# Save to WAV
engine.save_to_file(poem, 'poem.wav')
engine.runAndWait()

print("poem.wav has been created successfully.")
```
